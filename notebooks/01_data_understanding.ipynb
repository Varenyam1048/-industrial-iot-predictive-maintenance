# 01 — Data Understanding
Exploratory look at the raw factory sensor dataset before any cleaning or feature engineering.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/factory_sensor_simulator_2040.csv')
print('Dataset shape:', df.shape)
df.head()
```

```python
df.info()
print('Machine types:', df['Machine_Type'].nunique())
print(df['Failure_Within_7_Days'].value_counts())
```

```python
missing = df.isna().sum().sort_values(ascending=False)
print(missing[missing > 0])
```

```python
print('Negative vibration readings:', (df['Vibration_mms'] < 0).sum())
print('Negative power readings:', (df['Power_Consumption_kW'] < 0).sum())
print()
print('Failure rate by machine type (%):')
print((df.groupby('Machine_Type')['Failure_Within_7_Days'].mean() * 100).sort_values(ascending=False).round(2))
```

```python
df.describe(include='all').T
```

**Takeaways:** a handful of sensors (`Laser_Intensity`, `Hydraulic_Pressure_bar`, `Coolant_Flow_L_min`, `Heat_Index`) are only fitted to some machine types and are structurally missing rather than faulty. `Vibration_mms` and `Power_Consumption_kW` contain a small number of physically impossible negative readings. Both issues are handled in `02_data_preprocessing.ipynb` / `src/data_pipeline.py`.
